```python
import pandas as pd

transactions = pd.read_csv(
    "/content/drive/MyDrive/Revenue_Intelligence_Project/Data/transaction_analytics.csv"
)

transactions.head()
```

```python
transactions.info()
```

```python
# ==========================================
# Convert Invoice Date
# ==========================================

transactions["InvoiceDate"] = pd.to_datetime(
    transactions["InvoiceDate"]
)

transactions.info()
```

```python
# ==========================================
# Create Year-Month
# ==========================================

transactions["YearMonth"] = (
    transactions["InvoiceDate"]
    .dt.to_period("M")
    .astype(str)
)

transactions[
    [
        "InvoiceDate",
        "YearMonth"
    ]
].head()
```

```python
# ==========================================
# Monthly Business Summary
# ==========================================

monthly_summary = (
    transactions
    .groupby("YearMonth")
    .agg(
        TotalRevenue=("Revenue", "sum"),
        ActiveCustomers=("Customer ID", "nunique"),
        TotalOrders=("Invoice", "nunique"),
        TotalTransactions=("Invoice", "count")
    )
    .reset_index()
)

monthly_summary
```

```python
# ==========================================
# Save Monthly Summary
# ==========================================

monthly_summary.to_csv(
    "/content/drive/MyDrive/Revenue_Intelligence_Project/Data/monthly_business_summary.csv",
    index=False
)

print("Monthly Business Summary Saved Successfully!")
```

```python
# ==========================================
# Add Average Order Value
# ==========================================

monthly_summary["AverageOrderValue"] = (
    monthly_summary["TotalRevenue"] /
    monthly_summary["TotalOrders"]
).round(2)

monthly_summary.head()
```

```python
monthly_summary.to_csv(
    "/content/drive/MyDrive/Revenue_Intelligence_Project/Data/monthly_business_summary.csv",
    index=False
)

print("Monthly Business Summary Updated Successfully!")
```
