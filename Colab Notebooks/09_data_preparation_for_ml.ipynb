# Phase 7 - Data Preparation for Machine Learning

How can customer behaviour data be prepared for an accurate churn prediction model?

## Why is this important?

Machine Learning algorithms require clean, structured and numerical data.

## Expected Output

A final modelling dataset ready for XGBoost.

## Business Value

Ensures reliable model performance while maintaining data quality.

```python
import pandas as pd
import numpy as np

from google.colab import drive
drive.mount('/content/drive')

features = pd.read_csv(
"/content/drive/MyDrive/Revenue_Intelligence_Project/Data/model_dataset.csv"
)

features.head()
```

```python
features.isnull().sum()
```

```python
features["AverageDaysBetweenOrders"] = (
    features["AverageDaysBetweenOrders"]
    .fillna(0)
)
```

```python
features.isnull().sum()
```

```python
features["Churn"].value_counts()
```

```python
features["Churn"].value_counts(normalize=True)*100
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
segment_encoder = LabelEncoder()

features["SegmentEncoded"] = (
    segment_encoder.fit_transform(
        features["Segment"]
    )
)

features[
[
"Segment",
"SegmentEncoded"
]
].drop_duplicates().sort_values("SegmentEncoded")
```

```python
age_encoder = LabelEncoder()

features["CustomerAgeEncoded"] = (
    age_encoder.fit_transform(
        features["CustomerAgeCategory"]
    )
)

features[
[
"CustomerAgeCategory",
"CustomerAgeEncoded"
]
].drop_duplicates().sort_values("CustomerAgeEncoded")
```

```python
model_data = features[
    [
        "Recency",
        "Frequency",
        "Monetary",
        "AverageOrderValue",
        "CustomerLifetime",
        "AverageDaysBetweenOrders",
        "SegmentEncoded",
        "CustomerAgeEncoded",
        "Churn"
    ]
]

model_data.head()
```

```python
model_data.isnull().sum()
```

```python
model_data.to_csv(
"/content/drive/MyDrive/Revenue_Intelligence_Project/Data/model_ready_dataset.csv",
index=False
)

print("Model Ready Dataset Saved Successfully!")
```
