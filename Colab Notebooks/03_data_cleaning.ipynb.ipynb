```python
# =====================================
# PROJECT SETUP
# =====================================

import pandas as pd
import numpy as np

from google.colab import drive
drive.mount('/content/drive')

FILE_PATH = "/content/drive/MyDrive/Revenue_Intelligence_Project/Data/online_retail_II.xlsx"
```

```python
# =====================================
# LOAD DATA
# =====================================

df_2009 = pd.read_excel(
    FILE_PATH,
    sheet_name="Year 2009-2010"
)

df_2010 = pd.read_excel(
    FILE_PATH,
    sheet_name="Year 2010-2011"
)

df = pd.concat(
    [df_2009, df_2010],
    ignore_index=True
)
```

**Creating** **2** **datasets**:

File 1
transaction_analytics.csv
Contains:
Purchases
Returns

File 2
customer_analytics.csv
Contains:
Purchases only

```python
df_clean = df.drop_duplicates()

print(df_clean.shape)
```

Removing Duplicates

Because Duplicates are true duplicates.

```python
df_clean = df_clean[
    df_clean["Customer ID"].notna()
]

print(df_clean.shape)
```

Removing Missing Customer IDs

Without Customer ID:

No customer journey
No RFM
No churn

```python
df_clean[
    df_clean["Description"]
    .str.contains(
        "bad debt",
        case=False,
        na=False
    )
]
```

Remove Bad Debt Records

Because those are not real customer data

Note: Above we see no records because those are already cleared in before step where customer IDs are missing

```python
df_clean["Revenue"] = (
    df_clean["Quantity"] *
    df_clean["Price"]
)
```

```python
transaction_analytics = df_clean.copy()
```

This create a data set of transaction analytics

This keeps:

Positive Revenue
Negative Revenue
Returns
Cancellations

```python
customer_analytics = df_clean[
    df_clean["Quantity"] > 0
].copy()
```

This creates a data set of customer analytics

This keeps:

Purchases only

```python
print("Transaction Dataset Shape:")
print(transaction_analytics.shape)

print()

print("Customer Dataset Shape:")
print(customer_analytics.shape)
```

```python
transaction_analytics.to_csv(
    "/content/drive/MyDrive/Revenue_Intelligence_Project/Data/transaction_analytics.csv",
    index=False
)

customer_analytics.to_csv(
    "/content/drive/MyDrive/Revenue_Intelligence_Project/Data/customer_analytics.csv",
    index=False
)

print("Files Saved")
```
