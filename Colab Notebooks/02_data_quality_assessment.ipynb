```python
# =====================================
# PROJECT SETUP
# =====================================

import pandas as pd
import numpy as np

from google.colab import drive
drive.mount('/content/drive')

FILE_PATH = "/content/drive/MyDrive/Revenue_Intelligence_Project/Data/online_retail_II.xlsx"
```

```python
# =====================================
# LOAD DATA
# =====================================

df_2009 = pd.read_excel(
    FILE_PATH,
    sheet_name="Year 2009-2010"
)

df_2010 = pd.read_excel(
    FILE_PATH,
    sheet_name="Year 2010-2011"
)

df = pd.concat(
    [df_2009, df_2010],
    ignore_index=True
)
```

Data Quality Assessment

```python
missing_pct = (
    df.isnull()
      .sum()
      .sort_values(ascending=False)
      / len(df)
      * 100
)

print(missing_pct)
```

```python
duplicates = df.duplicated().sum()

print("Duplicate Rows:", duplicates)
```

```python
df["Revenue"] = (
    df["Quantity"] *
    df["Price"]
)

print(df["Revenue"].describe())
```

```python
# Top countries
print(
    df["Country"]
    .value_counts()
    .head(10)
)
```
